Описание признаков датасета:

id - int64

Уникальный идентификатор пациента

N_Days - int64

Количество дней наблюдения за пациентом

Drug - object

Тип препарата: "D-penicillamine" или "Placebo"

Age - int64

Возраст пациента в днях

Sex - object

Пол пациента: "M" (мужской), "F" (женский)

Ascites - object

Наличие асцита: "Y" (да), "N" (нет)

Hepatomegaly - object

Наличие гепатомегалии: "Y" (да), "N" (нет)

Spiders - object

Наличие сосудистых звездочек: "Y" (да), "N" (нет)

Edema - object

Наличие отеков: "Y" (да), "N" (нет), "S" (легкая форма)

Bilirubin - float64

Уровень билирубина в сыворотке крови

Cholesterol - float64

Уровень холестерина

Albumin - float64

Уровень альбумина

Copper - float64

Уровень меди в моче

Alk_Phos - float64

Уровень щелочной фосфатазы

SGOT - float64

Уровень аспартатаминотрансферазы

Tryglicerides - float64

Уровень триглицеридов

Platelets - float64

Количество тромбоцитов

Prothrombin - float64

Протромбиновое время

Stage - float64

Стадия заболевания (1-4)

Status - object

Статус пациента: "C" (censored), "D" (died), "CL" (censored due to liver transplantation)
